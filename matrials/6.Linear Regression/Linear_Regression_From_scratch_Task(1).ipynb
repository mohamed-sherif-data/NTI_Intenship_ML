```python
# implementing linear regression equation
def linear_regression(x,w,b):
  y = x * w + b
  return y
```

#**Cost Function**
cost function.png

```python
def compute_cost(x, y, w, b, linear_regression):

    m = x.shape[0]
    cost = 0

    for i in range(m):
        f_wb = linear_regression(x[i], w, b)
        cost = cost + (f_wb - y[i])**2
    total_cost = 1 / (2 * m) * cost

    return total_cost
```

# **Computing drevatives of Gradient Descent**
image.png

```python
def compute_gradient(x, y, w, b, linear_regression):

    m = x.shape[0]
    dj_dw = 0
    dj_db = 0

    for i in range(m):
        f_wb = linear_regression(x[i], w, b)
        dj_dw_i = (f_wb - y[i]) * x[i]
        dj_db_i = f_wb - y[i]
        dj_db += dj_db_i
        dj_dw += dj_dw_i
    dj_dw = dj_dw / m
    dj_db = dj_db / m

    return dj_dw, dj_db
```
