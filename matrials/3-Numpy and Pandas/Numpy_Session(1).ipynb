if you want to add elements of two lists together like [1, 2, 3] + [1, 2, 3] = [2, 4, 6], doing this with regular lists will not works, the two lists will be concatenated not added mathimatically -- that's why we have Numpy, we can implement a huge number of mathematics operations on lists with Numpy!

```python
list1 = [1, 2, 3]
addition = list1 + list1  # wrong way to add two lists
addition
```

```python
list1 = np.array([1, 2, 3])
list2 = np.array([1, 1, 1])
addition = list1 + list2  # right way to add two lists
addition
```

**Conditions Using where**

**np.where():** It returns indices where a condition is True, or it can be used to choose values based on a condition (like an if-else for arrays).

```python
a = np.array([10, 20, 30, 40, 50])
indices = np.where(a > 25)

print(indices)
print(a[indices])
a[0]
```

```python
booll =a < 30
a[booll]
```

```python
b = np.array([1, 2, 3, 4, 5])
result = np.where(b % 2 == 0, 'even', 'odd')
print(result)
```

**Excercise 2: Sales Data Analysis**
- Create a 4×7 array of random sales between $100 and $500.

- Apply a 10% tax to sales exceeding $300. (x1.10)

- Use np.where() to give a 5% discount to sales below $200. (x 0.95)

- Find the average sales for weekends (last 2 columns).

- Identify days with sales > $400 (boolean mask).




```python
sales = np.random.uniform(100,500, size=(4,7))
sales
```

```python
sales[sales > 300] *= 1.10
sales= np.where(sales < 200, sales*0.95, sales)
weekend = sales[:, -2:]
avg = np.mean(weekend)
print(avg)
```

```python
sales[sales >400]
```
