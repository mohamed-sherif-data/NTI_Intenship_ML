# Charts For Numerical Data

**1- Bar Chart:** How oftern each value occured?

```python
df['bedrooms'].value_counts()
```

```python
plt.bar(df['bedrooms'].value_counts().index, df['bedrooms'].value_counts())
plt.show()
```

```python
sns.countplot(x='bedrooms', hue= 'furnishingstatus',data=df)
plt.show()
```

**2- Histogram**: showing the distribution

```python
plt.hist(df['price'], bins=50)
plt.show()
```

**3- scatterplot:** plotting the relationship between 2 variables

```python
plt.scatter(df['area'], df['price'], marker='o', color='green')
plt.title('Area vs Price')
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()
```

```python
plt.scatter(df['area'], df['price'], marker='o', c=df['price'])
plt.title('Area vs Price')
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()
```

**4-Boxplot:** To visualize outliers

```python
plt.boxplot(df['price'])
plt.show()
```

```python
sns.boxplot(x='price', data=df)
plt.show()
```

# Charts For Categorical Values

**1- barchart** : what is the frequency of each value?

```python
furnishing_counts = df['furnishingstatus'].value_counts()
colors = ['blue', 'green', 'red']

bars = plt.bar(furnishing_counts.index, furnishing_counts.values, color=colors)

plt.xlabel('Furnishing Status')
plt.ylabel('Number of Houses')
plt.title('Number of Houses by Furnishing Status')

plt.legend(bars, furnishing_counts.index, title='Furnishing Status')

plt.show()
```

**2-Countplot**: also for frequency

```python
sns.countplot(x='furnishingstatus', data=df)
plt.show()
```

**3- Pie Chart**

```python
plt.pie(furnishing_counts, labels=furnishing_counts.index,  autopct='%1.1f%%')
plt.title('Furnishing Status Distribution')
```

# Visualizing a Categorical and a Quantitative Variable

Catplot in seaborn allow us to plot a categorical variable with a numerical one with different charts

```python
sns.catplot(x='furnishingstatus', y='price', data=df, kind='bar')
plt.show()
```

```python
sns.catplot(x='airconditioning', y='price', data=df, kind='box')
plt.show()
```
